```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END


class WasteState(TypedDict):
    waste: str
    classification: str
    reason: str
    disposal: str
```

```python
def classify_waste(state: WasteState):

    waste = state["waste"].lower()

    if any(word in waste for word in [
        "food", "fruit", "vegetable", "banana",
        "apple", "peel", "leaves", "organic"
    ]):
        category = "Wet Waste"

    elif any(word in waste for word in [
        "plastic", "bottle", "wrapper",
        "polythene", "plastic bag"
    ]):
        category = "Dry Waste"

    elif any(word in waste for word in [
        "paper", "cardboard", "newspaper",
        "book", "notebook"
    ]):
        category = "Paper Waste"

    elif any(word in waste for word in [
        "glass", "jar"
    ]):
        category = "Glass Waste"

    elif any(word in waste for word in [
        "battery", "mobile", "charger",
        "computer", "laptop", "electronic",
        "earphone"
    ]):
        category = "E-Waste"

    elif any(word in waste for word in [
        "medicine", "tablet", "syringe",
        "medical"
    ]):
        category = "Medical Waste"

    elif any(word in waste for word in [
        "metal", "iron", "steel",
        "aluminium", "copper", "can"
    ]):
        category = "Metal Waste"

    else:
        category = "General Waste"

    return {
        "classification": category
    }
```

```python
def explain_waste(state: WasteState):

    category = state["classification"]

    explanations = {
        "Wet Waste":
            "Biodegradable waste mainly obtained from food and organic materials.",

        "Dry Waste":
            "Non-biodegradable or recyclable waste such as plastic materials.",

        "Paper Waste":
            "Paper-based waste that can generally be recycled.",

        "Glass Waste":
            "Glass material that should be handled carefully and recycled.",

        "E-Waste":
            "Electronic waste that requires specialized recycling.",

        "Medical Waste":
            "Medical-related waste that requires appropriate handling.",

        "Metal Waste":
            "Metal materials that can generally be separated and recycled.",

        "General Waste":
            "The item could not be identified from the available categories."
    }

    return {
        "reason": explanations[category]
    }
```

```python
def disposal_advice(state: WasteState):

    category = state["classification"]

    disposal = {
        "Wet Waste":
            "Place it in the wet or organic waste collection stream.",

        "Dry Waste":
            "Keep it clean and dry and place it in the dry-waste stream.",

        "Paper Waste":
            "Keep it dry and send it for paper recycling.",

        "Glass Waste":
            "Handle carefully and send it to a glass-recycling collection system.",

        "E-Waste":
            "Take it to an authorized e-waste collection or recycling facility.",

        "Medical Waste":
            "Use an appropriate authorized medical-waste collection system.",

        "Metal Waste":
            "Separate it and send it for metal recycling.",

        "General Waste":
            "Follow your local waste-management instructions."
    }

    return {
        "disposal": disposal[category]
    }
```

```python
workflow = StateGraph(WasteState)

workflow.add_node("classify", classify_waste)
workflow.add_node("explain", explain_waste)
workflow.add_node("disposal", disposal_advice)

workflow.add_edge(START, "classify")
workflow.add_edge("classify", "explain")
workflow.add_edge("explain", "disposal")
workflow.add_edge("disposal", END)

app = workflow.compile()

print("LangGraph created successfully!")
```

```python
waste = input("Enter the waste item: ")

result = app.invoke({
    "waste": waste,
    "classification": "",
    "reason": "",
    "disposal": ""
})
```

```python
print("\n========== WASTE ANALYSIS ==========")

print("Waste Item:", waste)

print("\nClassification:")
print(result["classification"])

print("\nReason:")
print(result["reason"])

print("\nDisposal Guidance:")
print(result["disposal"])

print("\n====================================")
```
