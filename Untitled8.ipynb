```python
import os
from getpass import getpass

os.environ["GEMINI_API_KEY"] = getpass("Enter your Gemini API key: ")

from google import genai

client = genai.Client(
    api_key=os.environ["GEMINI_API_KEY"]
)

print("Gemini client created successfully!")
```

```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END


class WasteState(TypedDict):
    waste: str
    classification: str
    reason: str
    disposal: str
```

```python
import time


def ask_gemini(prompt):

    models = [
        "gemini-3.8-flash",
        "gemini-3.7-flash",
        "gemini-3.5-flash-lite"
    ]

    for model in models:

        for attempt in range(3):

            try:
                response = client.models.generate_content(
                    model=model,
                    contents=prompt
                )

                return response.text.strip()

            except Exception as e:

                error = str(e)

                if "503" in error or "UNAVAILABLE" in error:

                    wait_time = 2 ** attempt

                    print(
                        f"Temporary Gemini server error. "
                        f"Retrying in {wait_time} seconds..."
                    )

                    time.sleep(wait_time)

                else:
                    raise e

    return "Gemini service is temporarily unavailable. Please try again."
```

```python
def classify_waste(state: WasteState):

    prompt = f"""
You are a waste classification assistant.

Waste item:
{state["waste"]}

Choose exactly one category:

Wet Waste
Dry Waste
Paper Waste
Glass Waste
E-Waste
Metal Waste
Medical Waste
General Waste

Return only the category name.
"""

    result = ask_gemini(prompt)

    return {
        "classification": result
    }
```

```python
def explain_waste(state: WasteState):

    prompt = f"""
Waste item: {state["waste"]}

Category: {state["classification"]}

Explain in two simple sentences why this item
belongs to this category.
"""

    result = ask_gemini(prompt)

    return {
        "reason": result
    }
```

```python
def disposal_advice(state: WasteState):

    prompt = f"""
Waste item: {state["waste"]}

Category: {state["classification"]}

Give short, environmentally responsible disposal
or recycling guidance.

Keep the answer simple.
"""

    result = ask_gemini(prompt)

    return {
        "disposal": result
    }
```

```python
workflow = StateGraph(WasteState)

workflow.add_node("classification", classify_waste)
workflow.add_node("explanation", explain_waste)
workflow.add_node("disposal", disposal_advice)

workflow.add_edge(START, "classification")
workflow.add_edge("classification", "explanation")
workflow.add_edge("explanation", "disposal")
workflow.add_edge("disposal", END)

app = workflow.compile()

print("LangGraph created successfully!")
```

```python
waste = input("Enter the waste item: ")

result = app.invoke({
    "waste": waste,
    "classification": "",
    "reason": "",
    "disposal": ""
})
```

```python
print("\n" + "=" * 50)
print("       AI WASTE CLASSIFICATION RESULT")
print("=" * 50)

print("\nWaste Item:")
print(result["waste"])

print("\nClassification:")
print(result["classification"])

print("\nReason:")
print(result["reason"])

print("\nDisposal Guidance:")
print(result["disposal"])

print("\n" + "=" * 50)
```
